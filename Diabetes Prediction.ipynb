```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
data=pd.read_csv('/content/diabetes.csv')
```

```python
data.shape
```

```python
X=data.iloc[:,:-1]
Y=data.iloc[:,-1]


Xtrain,Xtest,Ytrain,Ytest=train_test_split(X,Y,test_size=0.25,stratify=Y,random_state=1)


scaler=StandardScaler()
Xtrain=scaler.fit_transform(Xtrain)
Xtest=scaler.transform(Xtest)


model=SVC(kernel='linear')
model.fit(Xtrain,Ytrain)


trainPredict=model.predict(Xtrain)
trainacc=accuracy_score(trainPredict,Ytrain)
print(trainacc)


testPredict=model.predict(Xtest)
testacc=accuracy_score(testPredict,Ytest)
print(testacc)


inp=(7,150,66,42,342,34.7,0.718,42)
converted=np.asarray(inp)
reshaped=converted.reshape(1,-1)
new=scaler.transform(reshaped)
Result=model.predict(new)
print(Result)
if(Result[0]==1):print("YESSSS")
else:print("NOOOO")
```
